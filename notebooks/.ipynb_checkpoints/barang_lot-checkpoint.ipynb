# SMART Barangs & Lots Staging & Final Master ETL
### Compilation of Subcategory Asset Partitions into Catalog Items (`barangs`), Procurement Batches (`lots`), and Unit Relational Linking (`lot_id`, `vendor_id`)
---
This notebook executes the ETL pipeline to map AMS assets and units into SMART:
1. **Catalog Deduction (`barangs`)**: Deduce unique item definitions across each subcategory by compiling units with matching `(subcategory, brand_id, name, specification)`.
2. **Procurement Batches (`lots`)**: Group units of each deduced barang into a single consolidated procurement lot (`LOT-0001-26-<barang.number>`).
3. **Relational Unit Linking (`units`)**: Link every unit to its parent `lot_id` and assign `vendor_id` from legacy AMS vendor mappings.
4. **Manifest & Data Exports**: Export subcategory partitions and consolidated master CSVs to `data/smart/staging/` and `data/smart/final/`, and generate SQL Server insert scripts.
5. **Data Quality & Integrity Assertions**: Run automated validation verifying monotonic sequential PKs, strict length constraints, and referential integrity.

```python
import os
import sys
import json
import re
from pathlib import Path
from collections import defaultdict, Counter
from datetime import datetime
import pandas as pd
import numpy as np

# ---------------------------------------------------------
# Dynamic Path Configurations
# ---------------------------------------------------------
def get_data_dir():
    candidates = [
        Path('/home/jovyan/data'),                      # Docker container
        Path('data'),                                  # Repository root
        Path('../data'),                               # Notebooks directory
        Path('/home/ran/Devs/ams-smart-migration/data') # Absolute host path
    ]
    for p in candidates:
        if p.exists() and (p / 'smart').exists():
            return p.resolve()
    return Path('data').resolve()

DATA_DIR = get_data_dir()
STAGING_ASSETS_DIR = DATA_DIR / 'smart' / 'staging' / 'assets'
STAGING_UNITS_DIR = DATA_DIR / 'smart' / 'staging' / 'units'
STAGING_BARANGS_DIR = DATA_DIR / 'smart' / 'staging' / 'barangs'
STAGING_LOTS_DIR = DATA_DIR / 'smart' / 'staging' / 'lots'

FINAL_BARANG_DIR = DATA_DIR / 'smart' / 'final' / 'barang'
FINAL_LOT_DIR = DATA_DIR / 'smart' / 'final' / 'lot'
FINAL_UNIT_DIR = DATA_DIR / 'smart' / 'final' / 'unit'

FINAL_SUBCATEGORY_PATH = DATA_DIR / 'smart' / 'final' / 'subcategory' / 'subcategories.csv'
FINAL_BRAND_PATH = DATA_DIR / 'smart' / 'final' / 'brand' / 'brands.csv'
FINAL_VENDOR_PATH = DATA_DIR / 'smart' / 'final' / 'vendor' / 'vendors.csv'

STAGING_BRAND_MAP_PATH = DATA_DIR / 'smart' / 'staging' / 'brand' / 'brand_ams_mapping.csv'
STAGING_VENDOR_MAP_PATH = DATA_DIR / 'smart' / 'staging' / 'vendor' / 'vendor_ams_mapping.csv'
UNITS_MANIFEST_PATH = STAGING_UNITS_DIR / '_units_manifest.json'

for d in [STAGING_BARANGS_DIR, STAGING_LOTS_DIR, FINAL_BARANG_DIR, FINAL_LOT_DIR, FINAL_UNIT_DIR]:
    d.mkdir(parents=True, exist_ok=True)

print(f'Data directory resolved: {DATA_DIR}')
```

## 2. Load Reference Master Datasets
We load target master tables and cross-reference dictionaries:
1. `subcategories.csv`: Maps `subcategory_code` to `subcategory_id`.
2. `brands.csv`: Target brand definitions and IDs.
3. `brand_ams_mapping.csv`: Mapped legacy AMS brands to SMART brand IDs.
4. `vendors.csv` & `vendor_ams_mapping.csv`: Legacy AMS vendors mapped to SMART vendor IDs.
5. `_units_manifest.json`: Manifest of subcategory partition files.

```python
# 1. Subcategory Lookup
df_sub = pd.read_csv(FINAL_SUBCATEGORY_PATH, dtype=str)
sub_map = dict(zip(df_sub['code'], df_sub['id']))
print(f'Loaded {len(sub_map)} subcategories.')

# 2. Brand Lookup & Aliases
df_brand = pd.read_csv(FINAL_BRAND_PATH, dtype=str)
brand_name_to_id = {r['name'].strip().upper(): str(r['id']).strip() for _, r in df_brand.iterrows() if pd.notna(r['name'])}

df_bmap = pd.read_csv(STAGING_BRAND_MAP_PATH, dtype=str)
ams_brand_to_id = {}
for _, r in df_bmap.iterrows():
    if pd.notna(r.get('ams_brand_name')):
        ams_brand_to_id[r['ams_brand_name'].strip().upper()] = str(r['smart_brand_id']).strip()

brand_aliases = {
    'LIONSTAR': brand_name_to_id.get('LION STAR', '125'),
    'LION STAR': brand_name_to_id.get('LION STAR', '125'),
    'PONGS': brand_name_to_id.get('PONGS HOME', '79'),
    'PONGS HOME': brand_name_to_id.get('PONGS HOME', '79'),
    'SYLYART': brand_name_to_id.get('SYLY ART', '130'),
    'SYLU ART': brand_name_to_id.get('SYLY ART', '130'),
    'SOFA': '123',
    'NO BRAND': '123',
    'CHITOSE': '173',
    'ICA': '174',
    'STELA': '175',
    'STELLA': '175',
    'MORELLY': '176',
}

# 3. Vendor Lookup
df_vmap = pd.read_csv(STAGING_VENDOR_MAP_PATH, dtype=str)
vendor_map = {}
for _, r in df_vmap.iterrows():
    v_id = str(r['ams_id']).strip()
    if v_id.endswith('.0'):
        v_id = v_id[:-2]
    vendor_map[v_id] = str(r['smart_vendor_id']).strip()

print(f'Loaded {len(brand_name_to_id)} brands and {len(vendor_map)} mapped vendors.')
```

## 3. Helper Functions Specification
We define transformation and resolution helper functions:
* **`clean_text` & `clean_spec`**: Whitespace collapse, carriage return removal, and empty/NULL placeholder handling.
* **`resolve_brand_id`**: Multi-tiered resolution prioritizing aliases, SMART master names, and AMS mappings, falling back to ID 123 (`No Brand`).
* **`resolve_vendor_id`**: Resolves AMS vendor ID to SMART vendor ID.
* **`resolve_organizer_id`**: Maps CFS $\rightarrow 1$ and ICT $\rightarrow 2$.
* **`pick_best_display_name` & `pick_best_display_spec`**: Deduce the cleanest, most representative name and specification within grouped units.

```python
def clean_text(s):
    if pd.isna(s) or s is None:
        return ''
    s = str(s).replace('\r', ' ').replace('\n', ' ')
    s = re.sub(r'\s+', ' ', s)
    return s.strip()

def clean_spec(s):
    s = clean_text(s)
    if s.upper() in ('', 'NULL', '-', '--', 'NONE', 'N/A'):
        return ''
    return s

def resolve_brand_id(raw_brand):
    if pd.isna(raw_brand) or not str(raw_brand).strip():
        return '123'  # No Brand
    b = str(raw_brand).strip().upper()
    if b in brand_aliases:
        return brand_aliases[b]
    if b in brand_name_to_id:
        return brand_name_to_id[b]
    if b in ams_brand_to_id:
        return ams_brand_to_id[b]
    return '123'

def resolve_vendor_id(raw_vendor):
    if pd.isna(raw_vendor) or str(raw_vendor).strip() in ('', 'NULL', 'null', 'None'):
        return None
    v = str(raw_vendor).strip()
    if v.endswith('.0'):
        v = v[:-2]
    return vendor_map.get(v, None)

def resolve_organizer_id(raw_org):
    org = str(raw_org).strip().upper() if pd.notna(raw_org) else ''
    return '2' if org == 'ICT' else '1'

def pick_best_display_name(raw_names):
    cleaned = [clean_text(n) for n in raw_names if clean_text(n)]
    if not cleaned:
        return 'Barang Tanpa Nama'
    counts = Counter(cleaned)
    return sorted(counts.keys(), key=lambda x: (counts[x], any(c.islower() for c in x)), reverse=True)[0]

def pick_best_display_spec(raw_specs):
    cleaned = [clean_spec(s) for s in raw_specs if clean_spec(s)]
    if not cleaned:
        return None
    counts = Counter(cleaned)
    return counts.most_common(1)[0][0]
```

## 4. Subcategory Partition Processing Loop
We iterate across all 111 subcategories in `_units_manifest.json`:
1. Load subcategory partitioned assets and cleaned units.
2. Compile units into unified barang records by `(brand_id, name_key, spec_key)`.
3. Assign sequential `number`: `{sub_code}-{incremental:04d}` for barangs and `LOT-0001-26-{barang.number}` for lots.
4. Set `uom_id = 1` ('Unit').
5. Link each unit to its assigned `lot_id` and map `vendor_id`.
6. Write partitioned CSV files into `staging/barangs/`, `staging/lots/`, and update `staging/units/`.

```python
with open(UNITS_MANIFEST_PATH, 'r', encoding='utf-8') as f:
    units_manifest = json.load(f)

current_barang_id = 1
current_lot_id = 1
now_str = datetime.now().strftime('%Y-%m-%d %H:%M:%S.000')

all_barangs_master = []
all_lots_master = []
all_units_updated_master = []

barangs_manifest = {'subcategories': {}, 'total_barangs': 0}
lots_manifest = {'subcategories': {}, 'total_lots': 0}
updated_units_manifest = {'subcategories': {}, 'total_units': 0}

for sub_code, meta in sorted(units_manifest.get('subcategories', {}).items()):
    unit_csv_path = STAGING_UNITS_DIR / meta['file']
    asset_csv_path = STAGING_ASSETS_DIR / meta['file']
    sub_barangs_csv = STAGING_BARANGS_DIR / meta['file']
    sub_lots_csv = STAGING_LOTS_DIR / meta['file']
    
    if not unit_csv_path.exists() or meta['row_count'] == 0:
        continue
        
    df_units = pd.read_csv(unit_csv_path, dtype=str)
    df_assets = pd.read_csv(asset_csv_path, dtype=str)
    
    df_merged = df_assets.merge(
        df_units[['ams_asset_id', 'image_url', 'created_at', 'updated_at']],
        left_on='id',
        right_on='ams_asset_id',
        suffixes=('_asset', '_unit')
    )
    
    groups = defaultdict(list)
    for _, r in df_merged.iterrows():
        b_id = resolve_brand_id(r.get('brand'))
        n_clean = clean_text(r.get('name'))
        s_clean = clean_spec(r.get('spesifikasi'))
        groups[(b_id, n_clean.upper(), s_clean.upper())].append(r)
        
    sorted_groups = sorted(groups.items(), key=lambda item: int(float(item[1][0]['id'])))
    
    sub_barangs = []
    sub_lots = []
    unit_lot_map = {}
    sub_cat_id = sub_map[sub_code]
    seq_counter = 1
    
    for (b_id, n_key, s_key), group_rows in sorted_groups:
        barang_num = f'{sub_code}-{seq_counter:04d}'
        lot_num = f'LOT-0001-26-{barang_num}'
        
        display_name = pick_best_display_name([r.get('name') for r in group_rows])
        display_spec = pick_best_display_spec([r.get('spesifikasi') for r in group_rows])
        
        first_img = None
        for r in group_rows:
            img = r.get('image_url') or r.get('image_url_unit')
            if pd.notna(img) and str(img).strip():
                first_img = str(img).strip()
                break
                
        created_ats = [str(r.get('created_at') or r.get('created_at_unit')) for r in group_rows if pd.notna(r.get('created_at') or r.get('created_at_unit')) and str(r.get('created_at') or r.get('created_at_unit')).strip()]
        updated_ats = [str(r.get('updated_at') or r.get('updated_at_unit')) for r in group_rows if pd.notna(r.get('updated_at') or r.get('updated_at_unit')) and str(r.get('updated_at') or r.get('updated_at_unit')).strip()]
        
        oldest_created = min(created_ats) if created_ats else now_str
        newest_updated = max(updated_ats) if updated_ats else now_str
        org_id = resolve_organizer_id(group_rows[0].get('organizer'))
        
        barang_row = {
            'id': current_barang_id,
            'number': barang_num,
            'subcategory_id': sub_cat_id,
            'brand_id': b_id,
            'uom_id': 1,
            'name': display_name,
            'specification': display_spec,
            'min_stock_threshold': None,
            'image_url': first_img,
            'last_restock_at': None,
            'created_at': oldest_created,
            'updated_at': newest_updated
        }
        
        lot_row = {
            'id': current_lot_id,
            'number': lot_num,
            'barang_id': current_barang_id,
            'organizer_id': int(org_id),
            'vendor_id': None,
            'legacy_vendor_id': None,
            'location_id': None,
            'initial_quantity': None,
            'current_quantity': None,
            'po_number': None,
            'date_of_receipt': now_str,
            'unit_price': None,
            'image_url': first_img,
            'burden': None,
            'project_id': None,
            'created_at': now_str,
            'updated_at': now_str
        }
        
        sub_barangs.append(barang_row)
        all_barangs_master.append(barang_row)
        sub_lots.append(lot_row)
        all_lots_master.append(lot_row)
        
        for r in group_rows:
            unit_lot_map[str(r['id']).strip()] = current_lot_id
            
        current_barang_id += 1
        current_lot_id += 1
        seq_counter += 1
        
    pd.DataFrame(sub_barangs).to_csv(sub_barangs_csv, index=False)
    pd.DataFrame(sub_lots).to_csv(sub_lots_csv, index=False)
    barangs_manifest['subcategories'][sub_code] = {'file': meta['file'], 'row_count': len(sub_barangs)}
    lots_manifest['subcategories'][sub_code] = {'file': meta['file'], 'row_count': len(sub_lots)}
    
    asset_vendor_map = {str(r['id']).strip(): resolve_vendor_id(r.get('vendor')) for _, r in df_assets.iterrows()}
    updated_units = []
    for _, u_row in df_units.iterrows():
        u_ams_id = str(u_row['ams_asset_id']).strip()
        assigned_lot_id = unit_lot_map.get(u_ams_id)
        assigned_vendor_id = asset_vendor_map.get(u_ams_id)
        
        u_dict = {
            'id': int(float(u_row['id'])),
            'ams_asset_id': u_ams_id,
            'number': u_row['number'],
            'legacy_number': u_row['legacy_number'],
            'lot_id': assigned_lot_id,
            'subcategory_code': u_row['subcategory_code'],
            'location_id': int(float(u_row['location_id'])),
            'status': u_row['status'],
            'condition': u_row['condition'],
            'price': float(u_row['price']) if pd.notna(u_row.get('price')) else None,
            'image_url': u_row['image_url'] if pd.notna(u_row.get('image_url')) else None,
            'ams_image_url': u_row['ams_image_url'] if pd.notna(u_row.get('ams_image_url')) else None,
            'type': u_row['type'],
            'classification': u_row['classification'],
            'burden': u_row['burden'],
            'project_id': int(float(u_row['project_id'])) if pd.notna(u_row.get('project_id')) and str(u_row.get('project_id')).strip() not in ('', 'None', 'nan') else None,
            'vendor_id': int(float(assigned_vendor_id)) if assigned_vendor_id else None,
            'created_at': u_row['created_at'],
            'updated_at': u_row['updated_at']
        }
        updated_units.append(u_dict)
        all_units_updated_master.append(u_dict)
        
    df_up_units = pd.DataFrame(updated_units)
    for c in ['id', 'lot_id', 'location_id', 'project_id', 'vendor_id']:
        if c in df_up_units.columns:
            df_up_units[c] = pd.to_numeric(df_up_units[c], errors='coerce').astype('Int64')
    df_up_units.to_csv(unit_csv_path, index=False)
    updated_units_manifest['subcategories'][sub_code] = {'file': meta['file'], 'row_count': len(df_up_units)}

barangs_manifest['total_barangs'] = len(all_barangs_master)
lots_manifest['total_lots'] = len(all_lots_master)
updated_units_manifest['total_units'] = len(all_units_updated_master)

print(f'Successfully compiled {len(all_barangs_master):,} barangs and {len(all_lots_master):,} lots from {len(all_units_updated_master):,} units.')
```

## 5. Consolidated Master Export & Manifest
We export the master files to staging and final production directories, and write SQL Server insert scripts.

```python
df_barangs_master = pd.DataFrame(all_barangs_master)
df_lots_master = pd.DataFrame(all_lots_master)
df_units_master = pd.DataFrame(all_units_updated_master)

for c in ['id', 'lot_id', 'location_id', 'project_id', 'vendor_id']:
    if c in df_units_master.columns:
        df_units_master[c] = pd.to_numeric(df_units_master[c], errors='coerce').astype('Int64')

# Staging Exports
df_barangs_master.to_csv(STAGING_BARANGS_DIR / 'barangs.csv', index=False)
with open(STAGING_BARANGS_DIR / '_barangs_manifest.json', 'w', encoding='utf-8') as f:
    json.dump(barangs_manifest, f, indent=2)

df_lots_master.to_csv(STAGING_LOTS_DIR / 'lots.csv', index=False)
with open(STAGING_LOTS_DIR / '_lots_manifest.json', 'w', encoding='utf-8') as f:
    json.dump(lots_manifest, f, indent=2)

df_units_master.to_csv(STAGING_UNITS_DIR / 'units.csv', index=False)
with open(STAGING_UNITS_DIR / '_units_manifest.json', 'w', encoding='utf-8') as f:
    json.dump(updated_units_manifest, f, indent=2)

# Final Production Exports
df_barangs_master.to_csv(FINAL_BARANG_DIR / 'barangs.csv', index=False)
df_lots_master.to_csv(FINAL_LOT_DIR / 'lots.csv', index=False)

db_ordered_cols = [
    'id', 'number', 'legacy_number', 'lot_id', 'location_id',
    'status', 'condition', 'price', 'image_url', 'vehicle_registration',
    'burden', 'project_id', 'created_at', 'updated_at', 'type',
    'classification', 'vendor_id'
]
df_units_final = df_units_master[[
    'id', 'number', 'legacy_number', 'lot_id', 'location_id',
    'status', 'condition', 'price', 'image_url', 'burden',
    'project_id', 'created_at', 'updated_at', 'type',
    'classification', 'vendor_id'
]].copy()
df_units_final['vehicle_registration'] = None
df_units_final = df_units_final[db_ordered_cols]
for c in ['id', 'lot_id', 'location_id', 'project_id', 'vendor_id']:
    df_units_final[c] = pd.to_numeric(df_units_final[c], errors='coerce').astype('Int64')
df_units_final.to_csv(FINAL_UNIT_DIR / 'units.csv', index=False)

print(f'Saved barangs: {len(df_barangs_master)} rows')
print(f'Saved lots:    {len(df_lots_master)} rows')
print(f'Saved units:   {len(df_units_final)} rows')
```

## 6. Data Quality & Referential Integrity Assertions
Rigorous automated validations verifying row counts, uniqueness, schema length limits, and foreign key references.

```python
# 1. Total row count assertion
assert len(df_units_master) == 6943, f'Expected 6,943 units, got {len(df_units_master)}'
assert len(df_barangs_master) == len(df_lots_master), 'Barangs count must match Lots count exactly'

# 2. Sequential continuous PK assertion
assert (df_barangs_master['id'] == list(range(1, len(df_barangs_master) + 1))).all(), 'Barangs PK must be monotonic 1..N'
assert (df_lots_master['id'] == list(range(1, len(df_lots_master) + 1))).all(), 'Lots PK must be monotonic 1..N'
assert (df_units_master['id'] == list(range(1, 6944))).all(), 'Units PK must be monotonic 1..6943'

# 3. Database column length limit assertions
max_b_len = df_barangs_master['number'].str.len().max()
max_l_len = df_lots_master['number'].str.len().max()
assert max_b_len <= 14, f'Barang number exceeds NVARCHAR(14): {max_b_len}'
assert max_l_len <= 26, f'Lot number exceeds NVARCHAR(26): {max_l_len}'

# 4. 100% Uniqueness
assert df_barangs_master['number'].is_unique, 'Barang number must be unique'
assert df_lots_master['number'].is_unique, 'Lot number must be unique'
assert df_units_master['number'].is_unique, 'Unit number must be unique'

# 5. Foreign Key Integrity
valid_sub_ids = set(df_sub['id'].astype(int))
assert set(df_barangs_master['subcategory_id'].astype(int)).issubset(valid_sub_ids), 'Invalid subcategory_id'

valid_brand_ids = set(df_brand['id'].astype(int))
assert set(df_barangs_master['brand_id'].astype(int)).issubset(valid_brand_ids), 'Invalid brand_id'

assert (df_barangs_master['uom_id'] == 1).all(), 'All uom_id must be 1 (Unit)'

valid_barang_ids = set(df_barangs_master['id'])
assert set(df_lots_master['barang_id']).issubset(valid_barang_ids), 'Invalid barang_id in lots'
assert set(df_lots_master['organizer_id']).issubset({1, 2}), 'Invalid organizer_id'

valid_lot_ids = set(df_lots_master['id'])
assert set(df_units_master['lot_id']).issubset(valid_lot_ids), 'Invalid lot_id in units'
assert df_units_master['lot_id'].notna().all(), 'lot_id cannot be null'

valid_vendor_ids = set(pd.read_csv(FINAL_VENDOR_PATH)['id'].astype(int))
non_null_vendors = set(df_units_master['vendor_id'].dropna().astype(int))
assert non_null_vendors.issubset(valid_vendor_ids), 'Invalid vendor_id in units'

# 6. Timestamp sanity
assert (df_barangs_master['created_at'] <= df_barangs_master['updated_at']).all(), 'created_at <= updated_at'

print('All 6 assertions passed successfully with 100% compliance!')
```

Inspect high-volume barangs, group distribution, and sample generated records.
