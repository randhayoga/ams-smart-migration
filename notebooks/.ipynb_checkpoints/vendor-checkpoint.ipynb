# Asset Vendor ETL & Data Cleaning
### AMS Legacy Migration to Smart Microsoft SQL Server (`vendors` table)

## 1. Overview & Business Context
This notebook migrates and cleans vendor master data from legacy AMS (`RE_ASSETS.dbo.tb_assets_vendor`) into the new Smart Enterprise Asset Management system (`smart.dbo.vendors`).

### Key Business Transformations:
1. **PT Writing Standardization**: Remove dots after `PT` and enforce standard Indonesian corporate naming conventions (`PT ` instead of `PT.` or `PT. `).
2. **Duplicate Code Separation**: Separate code collision `VN00080` (shared by `PT Danan Wingus Sakti` and `PT Berkat Niaga Dunia`), assigning `VN00118` to `PT Berkat Niaga Dunia` so no duplicated codes remain.
3. **Duplicate & Legacy Record Consolidation**:
   - Consolidate duplicate vendor `PT Roda Nurmala` (AMS ID 67 `VN00059` into AMS ID 57 `VN00049`).
   - Consolidate 6 early bare records (AMS IDs 13, 14, 15, 22, 26, 28) into their complete 2024 authoritative profiles.
4. **Product Suffix Extraction**: Move product category notes in parentheses (e.g. `(Roofing)`, `(Floordeck)`) into the `description` column.
5. **Data Cleaning & Normalization**:
   - Address: Clean newlines `\r\n` into `, `, allow `NULL` for missing addresses.
   - Phone Numbers: Strip leading quotes `'`, clean multi-line phones, resolve scientific notation in CP phone, allow `NULL` for missing numbers.
   - Emails: Clean whitespace and trailing delimiters, lowercase.
   - Contact Persons: Normalize names to Title Case, preserve phone leading zeros.
6. **Dual Export Pipeline**:
   - **Staging Outputs** (`data/smart/staging/vendor/`): `vendors.csv`, `vendor_ams_mapping.csv`, `relegated_vendors.csv`.
   - **Production Final Outputs** (`data/smart/final/vendor/`): `vendors.csv`, `vendors.sql`.

```python
import pandas as pd
import numpy as np
import re
from pathlib import Path
from datetime import datetime

# ---------------------------------------------------------
# Dynamic Path Configurations
# Robustly resolves data directory regardless of whether
# current working directory is /home/jovyan, /home/jovyan/work,
# or the repository root.
# ---------------------------------------------------------
def get_data_dir():
    candidates = [
        Path("data"),                                  # Running from project root
        Path("/home/jovyan/data"),                      # Running inside docker container
        Path("../data"),                               # Running from notebooks/ directory
        Path("/home/ran/Devs/ams-smart-migration/data") # Absolute host path
    ]
    for p in candidates:
        if p.exists() and (p / "ams").exists():
            return p.resolve()
    return Path("data").resolve()

DATA_DIR = get_data_dir()
RAW_DATA_PATH = DATA_DIR / "ams" / "tb_assets_vendor_202610011126.csv"
STAGING_DIR = DATA_DIR / "smart" / "staging" / "vendor"
FINAL_DIR = DATA_DIR / "smart" / "final" / "vendor"

STAGING_DIR.mkdir(parents=True, exist_ok=True)
FINAL_DIR.mkdir(parents=True, exist_ok=True)

print(f"Data directory:     {DATA_DIR}")
print(f"Raw AMS Vendor CSV: {RAW_DATA_PATH} (exists: {RAW_DATA_PATH.exists()})")
print(f"Staging directory:  {STAGING_DIR}")
print(f"Final directory:    {FINAL_DIR}")
```

```python
# Load raw vendor dataset, treating literal string 'NULL', 'null', and blanks as NaN
# Using dtype=str ensures phone numbers with leading zeros (e.g. '08...') are not corrupted to floats
df_raw = pd.read_csv(
    RAW_DATA_PATH,
    dtype=str,
    na_values=["NULL", "null", "NaN", "nan", ""],
    keep_default_na=True,
    encoding="utf-8"
)

print(f"Raw AMS vendor records loaded: {len(df_raw)} rows, {len(df_raw.columns)} columns")
display(df_raw.head(5))
```

## 3. Master Vendor Configuration, Cleaning Rules, & Casing Strategy
We apply systematic cleaning rules:
1. **Name Normalization & PT Writing**: Convert `PT.`/`P.T.` to `PT ` without dot, format to Title Case, preserve business acronyms (`PT`, `CV`, `PD`, `PCI`, `KSB`, `SAS`, `Tbk`, `SEA`, `USA`), and strip doubled parentheses.
2. **Product Suffix Extraction**: Move product notes (e.g. `(Roofing)`, `(Floordeck)`) into the `description` column.
3. **Duplicate Code Separation**: Assign `VN00118` to AMS ID 1095 (`PT Berkat Niaga Dunia`), while AMS ID 1087 (`PT Danan Wingus Sakti`) retains `VN00080`.
4. **Consolidation**:
   - Merge AMS ID 67 (`PT Roda Nurmala`) into AMS ID 57.
   - Merge 6 early bare records (AMS IDs 13, 14, 15, 22, 26, 28) into their 2024 complete profiles.
5. **Field Standardization**: Clean newlines in address/phone, remove leading quotes `'`, normalize emails, and format contact person details.

```python
# ---------------------------------------------------------
# 1. Product Suffixes Extraction Map
# Product category notes found in vendor name parentheses
# ---------------------------------------------------------
PRODUCT_SUFFIXES = {
    '47': 'Steel Door Bostinco',
    '48': 'Roofing',
    '49': 'Floordeck',
    '51': 'Steel Grating',
    '52': 'Spun Pile',
}

# ---------------------------------------------------------
# 2. Explicit Cleaned Name Map (118 verified corporate names)
# Enforces PT without dot, Title Case, and preserved acronyms
# ---------------------------------------------------------
EXPLICIT_NAME_MAP = {
    '9': 'Koperasi RE',
    '10': 'Alfa Solusi',
    '11': 'Cahaya Sukses',
    '12': 'Cascasia',
    '13': 'Cipta Satria Informatika',
    '14': 'Codeware',
    '15': 'Danan Wingus',
    '16': 'Daniel',
    '17': 'EDPMEDIA',
    '18': 'Everest Technology',
    '19': 'Glodok',
    '20': 'Gramedia',
    '21': 'Indosystem',
    '22': 'Multi Kharisma',
    '23': 'Multimedia Sejahtera Servitama',
    '24': 'Optima Solusindo Informatika',
    '25': 'Panca Putra',
    '26': 'Piranti Sarana',
    '27': 'Sarana Aircon Utama',
    '28': 'Sarana Solusindo',
    '29': 'Sentra Grafika Kompumedia',
    '30': 'Siemens',
    '31': 'Sinergi',
    '32': 'Surabaya',
    '33': 'Wilerdtindo',
    '34': 'PT Paradise Perkasa',
    '35': 'PT Loka Manggala',
    '36': 'PT Golden Pratama Engineering',
    '37': 'PT Crane Indonesia',
    '38': 'PT Mitra Galperti',
    '39': 'PT Fajar Benua Indopack',
    '40': 'PT Nichias Sunijaya',
    '41': 'PT Gamako Ekakarsa',
    '42': 'PT Binder Indonesia',
    '43': 'PT Kranindo Prima',
    '44': 'PT Karya Prima Suplindo',
    '45': 'PT KPM Oil & Gas',
    '46': 'PT Valvindo Megah',
    '47': 'PT Fors Fortid Indonesia',
    '48': 'PT Bluescope Lysaght Indonesia',
    '49': 'PT Union Metal',
    '50': 'Indonesian Epoxy Floor',
    '51': 'PT Karlita Emas',
    '52': 'Adimix PCI Indonesia',
    '53': 'PT KSB Indonesia & PT KSB Sales Indonesia',
    '54': 'PT Petrotec Guna Perkasa',
    '55': 'PT Ashcrofindo Manunggal',
    '56': 'PT Bumi Cahaya Unggul',
    '57': 'PT Roda Nurmala',
    '58': 'PT Duraquipt Cemerlang',
    '59': 'PT Enerflex Indonesia',
    '60': 'PT SAS International',
    '61': 'PT Fajar Mas Murni',
    '62': 'PT Pacific Process Engineering',
    '63': 'PT Dinar Inti Duba',
    '64': 'PT Indobara',
    '65': 'PT Rantai Laut',
    '66': 'PT Sabrindo',
    '67': 'PT Roda Nurmala',
    '68': 'PT Kalimutu Mitra Perkasa',
    '69': 'PT Rejeki Utama',
    '70': 'Trafindo',
    '71': 'Perpro',
    '72': 'Philips',
    '73': 'Kotaminyak',
    '74': 'Erico',
    '75': 'Schneider Electric',
    '76': 'Socomec',
    '77': 'Trakindo',
    '78': 'Berkah Cipta Persada',
    '79': 'Sucaco',
    '80': 'Koperasi Reka Teknika Utama',
    '1080': 'Kontraktor Rifai',
    '1081': 'PT Intisari Komputindo',
    '1082': 'Yolori',
    '1083': 'Protech Computer',
    '1084': 'Trinity Official Store',
    '1085': 'Agres ID',
    '1086': 'Trans Survey',
    '1087': 'PT Danan Wingus Sakti',
    '1088': 'PT Multi Prima Mandiri Sukses',
    '1089': 'PT Putraduta Buanasentosa (Indoarsip)',
    '1090': 'PD Duta Copy Service',
    '1091': 'Koperasi Reka Sejahtera',
    '1092': 'PT Sarana Solusindo Informatika',
    '1093': 'PT Abadi Jaya Makmur',
    '1094': 'PT Onna Prima Utama',
    '1095': 'PT Berkat Niaga Dunia',
    '1096': 'PT Berca Carrier Indonesia',
    '1097': 'PT Mitsubishi Jaya Elevator & Escalator',
    '1098': 'Codeware, Inc.',
    '1099': 'Bentley Systems International Ltd.',
    '1100': 'PT Lima Pilar Cakrawala',
    '1101': 'PT Cipta Satria Informatika',
    '1102': 'PT Medatama (Mega Persada Utama)',
    '1103': 'PT Recare',
    '1104': 'Trimble Solutions SEA Pte. Ltd.',
    '1105': 'PT Parsaoran Global Datatrans',
    '1106': 'Kayreach System',
    '1107': 'Cahaya Jaya Services',
    '1108': 'PT Piranti Sarana Technologi',
    '1109': 'PT Multi Kharisma',
    '1110': 'PT Lippo Insurance Tbk',
    '1111': 'PT Mitralab Buana',
    '1112': 'PT Arya Sentra Solusindo',
    '1113': 'PT Praba Sinergi Utama',
    '1114': 'Anril Film',
    '1115': 'Beat Studio',
    '1116': 'PT Prismaco',
    '1117': 'PT Kelola Buana Lestari',
    '1118': 'Intact Official Store',
    '1119': 'Tokopedia',
    '1120': 'Titis',
    '1121': 'Sunio Citra Teknologi',
    '1122': 'Sarana Solusindo Prima',
    '1123': 'PT Cipta Tekno Solusindo',
    '1124': 'PT United Teknologi Integrasi',
    '1125': 'PT Sparta Computindo Teknologi',
}

# ---------------------------------------------------------
# 3. Consolidation Mapping (Option A)
# Maps duplicate or bare legacy records to authoritative 2024 profiles
# ---------------------------------------------------------
MERGE_MAP = {
    '67': '57',    # PT Roda Nurmala (VN00059) -> PT Roda Nurmala (VN00049)
    '13': '1101',  # CIPTA SATRIA INFORMATIKA (VN00005) -> PT Cipta Satria Informatika (VN00093)
    '14': '1098',  # CODEWARE (VN00006) -> Codeware, Inc. (VN00090)
    '15': '1087',  # DANAN WINGUS (VN00007) -> PT Danan Wingus Sakti (VN00080)
    '22': '1109',  # MULTI KHARISMA (VN00014) -> PT Multi Kharisma (VN00101)
    '26': '1108',  # PIRANTI SARANA (VN00018) -> PT Piranti Sarana Technologi (VN00100)
    '28': '1092',  # SARANA SOLUSINDO (VN00020) -> PT Sarana Solusindo Informatika (VN00085)
}

# Helper cleaning functions
def clean_address(val):
    if pd.isna(val) or val is None or str(val).strip() == '':
        return None
    val = str(val).strip()
    val = re.sub(r'[\r\n]+', ', ', val)
    val = re.sub(r'\s+', ' ', val)
    val = re.sub(r'\s*,\s*', ', ', val)
    val = re.sub(r'(,\s*)+', ', ', val)
    val = val.strip(' ,')
    return val if val else None

def clean_phone(val):
    if pd.isna(val) or val is None or str(val).strip() == '':
        return None
    val = str(val).strip()
    val = val.lstrip("'")
    val = re.sub(r'[\r\n]+', ' / ', val)
    val = re.sub(r'\s*,\s*', ' / ', val)
    val = re.sub(r'\s+', ' ', val)
    val = re.sub(r'\s*/\s*', ' / ', val)
    val = val.strip(' /')
    return val if val else None

def clean_email(val):
    if pd.isna(val) or val is None or str(val).strip() == '':
        return None
    val = str(val).strip().rstrip(';,').strip().lower()
    return val if val else None

def clean_cp_name(val):
    if pd.isna(val) or val is None or str(val).strip() == '':
        return None
    val = str(val).strip()
    val = re.sub(r'Mrs,', 'Mrs.', val)
    val = re.sub(r'^(Mr|Mrs|Ms)\.([A-Za-z])', r'\1. \2', val)
    val = re.sub(r'([A-Za-z]\.)([A-Za-z])', r'\1 \2', val)
    words = val.split(' ')
    clean_words = []
    for w in words:
        if not w:
            continue
        w_clean = re.sub(r'[^A-Za-z]', '', w).upper()
        if w_clean in {'SE', 'MM', 'SH', 'ST', 'SIK', 'KOM'}:
            clean_words.append(w.upper())
        elif w.startswith('Mr.') or w.startswith('Mrs.') or w.startswith('Ms.') or w.startswith('Ir.'):
            clean_words.append(w[:1].upper() + w[1:])
        elif '.' in w:
            parts = w.split('.')
            clean_parts = [p.capitalize() for p in parts]
            clean_words.append('.'.join(clean_parts))
        else:
            clean_words.append(w.capitalize())
    return ' '.join(clean_words)

def clean_cp_phone(val, row_id):
    if pd.isna(val) or val is None or str(val).strip() == '':
        return None
    val = str(val).strip().lstrip("'")
    # Correct scientific notation export from legacy AMS for id 9
    if val == '2.17918e+009' and str(row_id) == '9':
        return '02179181180'
    return val if val else None

def clean_description(val, row_id):
    suffix = PRODUCT_SUFFIXES.get(str(row_id))
    if pd.isna(val) or val is None or str(val).strip() == '':
        base_desc = None
    else:
        base_desc = str(val).strip()
        base_desc = re.sub(r'[\r\n]+', ', ', base_desc)
        base_desc = re.sub(r'\s+', ' ', base_desc).strip()
    
    if suffix and base_desc:
        return f"{base_desc}, {suffix}"
    elif suffix:
        return suffix
    return base_desc

# Build transformed working dataset
work_rows = []
for idx, r in df_raw.iterrows():
    aid = str(r['id'])
    raw_code = str(r['code']).strip()
    
    # 1. Duplicate Code Separation: VN00080 for id 1095 becomes VN00118
    if aid == '1095' and raw_code == 'VN00080':
        code = 'VN00118'
    else:
        code = raw_code
    
    name = EXPLICIT_NAME_MAP[aid]
    address = clean_address(r['address'])
    phone_number = clean_phone(r['telfon'])
    email = clean_email(r['email'])
    description = clean_description(r['description'], aid)
    
    cp1_name = clean_cp_name(r['cp_name1'])
    cp1_tel = clean_cp_phone(r['cp_telp1'], aid)
    cp1_email = clean_email(r['cp_email1'])
    
    cp2_name = clean_cp_name(r['cp_name2'])
    cp2_tel = clean_cp_phone(r['cp_telp2'], aid)
    cp2_email = clean_email(r['cp_email2'])
    
    created_at = pd.to_datetime(r['create_date']).strftime('%Y-%m-%d %H:%M:%S')
    updated_at = created_at
    
    is_consolidated = aid in MERGE_MAP
    target_ams_id = MERGE_MAP.get(aid)
    
    work_rows.append({
        'ams_id': int(aid),
        'code': code,
        'name': name,
        'address': address,
        'phone_number': phone_number,
        'email': email,
        'description': description,
        'contact_person_1': cp1_name,
        'cp_email_1': cp1_email,
        'cp_phone_1': cp1_tel,
        'contact_person_2': cp2_name,
        'cp_email_2': cp2_email,
        'cp_phone_2': cp2_tel,
        'created_at': created_at,
        'updated_at': updated_at,
        'is_consolidated': is_consolidated,
        'target_ams_id': int(target_ams_id) if target_ams_id else None
    })

df_work = pd.DataFrame(work_rows)

# Build Production Final DataFrame (is_consolidated == False)
df_final = df_work[df_work['is_consolidated'] == False][
    ['code', 'name', 'address', 'phone_number', 'email', 'description', 
     'contact_person_1', 'cp_email_1', 'cp_phone_1', 'contact_person_2', 
     'cp_email_2', 'cp_phone_2', 'created_at', 'updated_at', 'ams_id']
].copy().reset_index(drop=True)

# Re-index final production IDs sequentially 1..N
df_final['id'] = range(1, len(df_final) + 1)

# Lookup mappings from ams_id to smart_id
smart_id_map = {}
smart_code_map = {}
smart_name_map = {}

for _, row in df_final.iterrows():
    smart_id_map[row['ams_id']] = row['id']
    smart_code_map[row['ams_id']] = row['code']
    smart_name_map[row['ams_id']] = row['name']

# Final columns alignment for production table
prod_cols = [
    'id', 'code', 'name', 'address', 'phone_number', 'email', 'description',
    'contact_person_1', 'cp_email_1', 'cp_phone_1', 'contact_person_2',
    'cp_email_2', 'cp_phone_2', 'created_at', 'updated_at'
]
df_final = df_final[prod_cols]

# Build 1-to-1 relational mapping table (df_mapping) and audit table (df_relegated)
mapping_rows = []
relegated_rows = []

for _, r in df_work.iterrows():
    aid = r['ams_id']
    raw_r = df_raw[df_raw['id'] == str(aid)].iloc[0]
    
    if r['is_consolidated']:
        tgt_aid = r['target_ams_id']
        smart_id = smart_id_map[tgt_aid]
        smart_code = smart_code_map[tgt_aid]
        smart_name = smart_name_map[tgt_aid]
        issue_type = 'duplicate_vendor' if aid == 67 else 'legacy_bare_record'
        action = f'Consolidated into {smart_name} (Smart ID: {smart_id}, Code: {smart_code})'
        note = f'Consolidated into AMS ID {tgt_aid}'
        
        relegated_rows.append({
            'ams_id': aid,
            'ams_code': raw_r['code'],
            'ams_vendor_name': raw_r['name'].strip(),
            'target_ams_id': tgt_aid,
            'target_smart_id': smart_id,
            'target_smart_code': smart_code,
            'target_smart_name': smart_name,
            'issue_type': issue_type,
            'action_taken': action,
            'notes': 'Merged into authoritative master profile'
        })
    else:
        smart_id = smart_id_map[aid]
        smart_code = r['code']
        smart_name = r['name']
        if aid == 1095:
            note = 'Reassigned code from VN00080 to VN00118 to resolve duplicate collision'
        else:
            note = 'Primary master record'
            
    mapping_rows.append({
        'ams_id': aid,
        'ams_code': raw_r['code'],
        'ams_vendor_name': raw_r['name'].strip(),
        'smart_vendor_id': smart_id,
        'smart_code': smart_code,
        'smart_vendor_name': smart_name,
        'is_consolidated': r['is_consolidated'],
        'notes': note
    })

df_mapping = pd.DataFrame(mapping_rows)
df_relegated = pd.DataFrame(relegated_rows)

# Build Staging Master DataFrame (df_staging)
# Groups mapped ams_ids for master records
staging_rows = []
for _, prod_row in df_final.iterrows():
    sid = prod_row['id']
    mapped_aids = df_mapping[df_mapping['smart_vendor_id'] == sid]['ams_id'].tolist()
    st_row = prod_row.to_dict()
    st_row['ams_id'] = str(mapped_aids)
    st_row['mapped_count'] = len(mapped_aids)
    st_row['has_consolidated_legacy'] = len(mapped_aids) > 1
    staging_rows.append(st_row)
df_staging = pd.DataFrame(staging_rows)

print(f"Staging master vendors prepared: {len(df_staging)} rows")
print(f"1-to-1 Mapping rows (all AMS):   {len(df_mapping)} rows")
print(f"Relegated/Consolidated rows:     {len(df_relegated)} rows")
print(f"Production Final vendors:        {len(df_final)} clean verified rows")
display(df_final.head(10))
```

## 4. Data Validation & Quality Checks
We execute rigorous automated assertions to verify data integrity:
1. All vendor codes are strictly unique and $\le 7$ characters (`VARCHAR(7)`).
2. All vendor names are strictly unique and $\le 255$ characters (`VARCHAR(255)`).
3. Zero occurrences of `PT.` or `P.T.` in any vendor name.
4. ID sequence starts at 1, increases monotonically without gaps.
5. All string columns satisfy $\le 255$ characters.
6. Exactly 118 original legacy AMS rows are accounted for in `df_mapping`.

```python
# 1. Check code uniqueness and length (VARCHAR(7))
assert df_final['code'].duplicated().sum() == 0, f"Error: Duplicate vendor codes detected: {df_final[df_final['code'].duplicated(keep=False)]['code'].tolist()}"
max_code_len = df_final['code'].str.len().max()
assert max_code_len <= 7, f"Error: Code length exceeds 7 characters ({max_code_len})"

# 2. Check name uniqueness and length (VARCHAR(255))
duplicate_names = df_final[df_final['name'].duplicated(keep=False)]
assert duplicate_names.empty, f"Error: Duplicate vendor names detected in final dataset: {duplicate_names['name'].tolist()}"
max_name_len = df_final['name'].str.len().max()
assert max_name_len <= 255, f"Error: Maximum vendor name length exceeds 255 characters ({max_name_len})"

# 3. Verify PT dot removal
assert not df_final['name'].str.contains(r'PT\.').any(), "Error: Found 'PT.' with dot in production names!"
assert not df_final['name'].str.contains(r'P\.T\.').any(), "Error: Found 'P.T.' with dots in production names!"

# 4. Check ID uniqueness and sequential continuity
assert df_final['id'].is_monotonic_increasing, "Error: IDs are not monotonic increasing"
assert df_final['id'].duplicated().sum() == 0, "Error: Duplicate IDs found in final dataset"
assert df_final['id'].iloc[0] == 1, "Error: ID sequence does not start at 1"
assert df_final['id'].iloc[-1] == len(df_final), "Error: ID sequence has gaps"

# 5. Check string lengths for all VARCHAR(255) columns
varchar_cols = ['address', 'phone_number', 'email', 'description', 'contact_person_1', 'cp_email_1', 'cp_phone_1', 'contact_person_2', 'cp_email_2', 'cp_phone_2']
for col in varchar_cols:
    max_len = df_final[col].dropna().astype(str).str.len().max()
    if pd.notna(max_len):
        assert max_len <= 255, f"Error: Column '{col}' exceeds 255 characters ({max_len})"

# 6. Check that all 118 original AMS rows are mapped
assert len(df_mapping) == len(df_raw), f"Error: Mapping table length ({len(df_mapping)}) does not match raw AMS rows ({len(df_raw)})"

print("All 6 data validation and integrity assertions PASSED successfully!")
```

   - `vendors.sql`: Executable MS SQL Server script with `IDENTITY_INSERT`.

```python
# ---------------------------------------------------------
# Export Production Outputs to data/smart/final/vendor/
# ---------------------------------------------------------
final_csv_path = FINAL_DIR / "vendors.csv"
final_sql_path = FINAL_DIR / "vendors.sql"

# 1. Target CSV for SQL Server Import (Strict 15 Columns)
df_final.to_csv(final_csv_path, index=False)

# Helper to format SQL literals
def sql_literal(val, is_unicode=True):
    if pd.isna(val) or val is None or str(val).strip() == '':
        return 'NULL'
    escaped = str(val).replace("'", "''")
    return f"N'{escaped}'" if is_unicode else f"'{escaped}'"

# 2. Generate SQL Server INSERT Script
sql_statements = [
    "-- ========================================================",
    "-- Master Vendors Data Migration Script for MS SQL Server",
    f"-- Generated: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}",
    "-- Target Table: vendors",
    "-- ========================================================",
    "SET IDENTITY_INSERT [vendors] ON;",
    "GO",
    ""
]

for _, row in df_final.iterrows():
    row_id = row['id']
    code_val = sql_literal(row['code'], is_unicode=False)
    name_val = sql_literal(row['name'], is_unicode=True)
    addr_val = sql_literal(row['address'], is_unicode=True)
    phone_val = sql_literal(row['phone_number'], is_unicode=True)
    email_val = sql_literal(row['email'], is_unicode=False)
    desc_val = sql_literal(row['description'], is_unicode=True)
    cp1_name_val = sql_literal(row['contact_person_1'], is_unicode=True)
    cp1_email_val = sql_literal(row['cp_email_1'], is_unicode=False)
    cp1_phone_val = sql_literal(row['cp_phone_1'], is_unicode=True)
    cp2_name_val = sql_literal(row['contact_person_2'], is_unicode=True)
    cp2_email_val = sql_literal(row['cp_email_2'], is_unicode=False)
    cp2_phone_val = sql_literal(row['cp_phone_2'], is_unicode=True)
    created_val = f"'{row['created_at']}'"
    updated_val = f"'{row['updated_at']}'"
    
    stmt = (
        f"INSERT INTO [vendors] ([id], [code], [name], [address], [phone_number], [email], [description], "
        f"[contact_person_1], [cp_email_1], [cp_phone_1], [contact_person_2], [cp_email_2], [cp_phone_2], "
        f"[created_at], [updated_at]) "
        f"VALUES ({row_id}, {code_val}, {name_val}, {addr_val}, {phone_val}, {email_val}, {desc_val}, "
        f"{cp1_name_val}, {cp1_email_val}, {cp1_phone_val}, {cp2_name_val}, {cp2_email_val}, {cp2_phone_val}, "
        f"{created_val}, {updated_val});"
    )
    sql_statements.append(stmt)

sql_statements.extend([
    "",
    "GO",
    "SET IDENTITY_INSERT [vendors] OFF;",
    "GO"
])

with open(final_sql_path, "w", encoding="utf-8") as f:
    f.write("\n".join(sql_statements))

print(f"Production files successfully exported to {FINAL_DIR.resolve()}:")
print(f" 1. {final_csv_path.name} ({len(df_final)} rows)")
print(f" 2. {final_sql_path.name} ({len(df_final)} INSERT statements)")
```
